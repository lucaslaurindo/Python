#Estrutura de Repetição com FOR

```python
nomes = ['Ana', 'José', 'Maria ']
for i in nomes:
  print('Olá, Seja bem vindo', i)
```

```python
for i in range(1,6):
  print(i)
```

```python
for i in range(10,0,-1):
  print(i)
```

```python
for i in range(0,10+1,2):
  print(i)
```

```python
numeros = [4,7,10,15,22,3]
for i in numeros:
  numeros.sort()
  if i % 2 == 0:
    print(i, 'é par')
  else:
    print(i, 'é ímpar')
```

```python
for i in range(11):
  if i % 2 == 0:
    print(i, 'é par')
  else:
    print(i, 'é ímpar')
```

```python
notas = [8.5, 9.2, 5.0, 3.0]
for nota in notas:
  if nota >= 6.0 and nota <=10.0:
    print(nota, 'Aprovado')
  else:
    print(nota, 'Reprovado')
```

```python
numeros = [4, 7, 10, 15, 22, 3]
soma = 0
for numero in numeros:
  soma = soma + numero
print('A soma é', soma)
print('A média é', soma/len(numeros))
```

```python
for numero in range(1,12):
  if numero == 8:
    break
  print(numero)
```

#***ATIVIDADE LISTAS DE*** ***REPETIÇÃO***

```python
#### Listas questão 01

for numero in range(1,21):
  print(numero)
```

```python
#### Listas questão 02
nomes = ['Ana', 'Bruno', 'Carla', 'Diego ']
for i in nomes:
  print('Olá, ', i)
```

```python
for i in range(1,31):
  if i % 2 == 0:
    print(i, 'é par')
  else:
    print(i, 'é ímpar')
```

```python
numeros = [3, 12, 7, 20, 5, 18, 9]
soma = 0
for i in numeros:
  if i >= 10:
    soma = soma + i
print(soma)
```

```python
#atividade 05
idades = [15, 22, 17, 30, 12, 19, 45]
contagem = 0
for i in idades:
  if i >= 18 and i <= 60:
    contagem += 1
print(contagem)
```

```python
notas = []
soma = 0
for i in range(5):
  nota = float(input("Digite sua nota: "))
  notas.append(nota)
print(notas)

for n in notas:
  soma = soma + n

media = soma / len(notas)
print(media)
if media >= 7 :
  print("A média está boa")
else:
  print("Precisa melhorar")
```

```python
def mensagem (nome):
  print(f'Bom dia, {nome} ')

mensagem('Ana')
mensagem('José')
```

```python
def apresentar(nome, idade):
  print(f'{nome}, tem {idade} anos')
apresentar('Carlos', 31)
apresentar('Ana', 21)
```

```python
##  parâmetro com valor padrão

def saudacao(nome, periodo='dia'):
  print(f'Bom {periodo}, {nome}')
saudacao('Ana')
saudacao('Carlos', 'noite')
```

```python
def somar(a, b):
  resultado = a + b
  return resultado
print(somar(105,20))
```

```python
def eh_par(numero):
  return numero % 2 == 0
print(eh_par(4))
print(eh_par(7))

num = int(input('Digite um número:'))
if eh_par(num):
  print('É par')
else:
  print('É impar')
```

```python
def calcular_media(notas):
  soma = 0
  for nota in notas:
    soma = soma + nota
  return soma / len(notas)
notas_a = [7,8,6]
notas_b = [5,9,5]

print(f'{calcular_media(notas_a):.2f}');
print(f'{calcular_media(notas_b):.2f}');
```

```python
####validador de idade
def maior_de_idade(idade):
  return idade >= 18
idade = int(input('Digite sua idade:'))
if maior_de_idade(idade):
  print('Acesso permitido')
else:
  ('Acesso bloqueado')
```

```python
#### calculadora de desconto
def calcular (preco, desconto_percentual):
  desconto = preco * (desconto_percentual/100)
  return preco - desconto
produto = int(input('Digite o valor do produto '))
desc = int(input('Digite o desconto '))
preco_final = calcular(produto,desc)
print(f'Preço final R$ {preco_final}')
```

```python
### atividade 01

def verificar_aprovacao(nota):
  if nota >= 6:
    return 'Aprovado'
  else:
    return 'Reprovado'
nota = float(input('Digite sua nota: '))
resultado = verificar_aprovacao(nota)
print(f'Você está {resultado}')
```

```python
### atividade 02

def soma_lista(numeros):
  soma= 0
  for numero in numeros:
    soma = soma + numero
  return soma
numeros = [10, 20, 30, 40]
print(soma_lista(numeros))
```

```python
### atividade 03

def filtrar_maiores_de_idade(idades):
  maiores = []
  for idade in idades:
    if idade >= 18:
      maiores.append(idade)
  return maiores
idades = [15, 18, 20, 16, 25]
print(filtrar_maiores_de_idade(idades))
```

```python
### atividade 04

def calcular_juros(valor, taxa=0.05):
  juros = valor * taxa
  return valor + juros
print(calcular_juros(1000))
print(calcular_juros(1000, 0.10))
```
